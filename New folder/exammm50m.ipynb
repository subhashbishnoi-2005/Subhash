```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("project.csv.csv")
print(df)
```

```python
#Task 1 Notebook Structure & Data Loading
print("DataSet Shape:\n",df.shape)
print("First Five Rows:\n",df.head(5))
print("Last Five Rows:\n",df.tail(5))
print("describe:\n",df.describe())

```

1.    first loaded the csv file by using ".read_csv"
2.    then get the shape of data set by using".shape"
3.    get the first five record by using ".head()"
4.    get the last five record by using ".tail()"

```python
#   Task 2: Data Cleaning & EDA
print("Missing Values:\n",df.isnull().sum())
dropNa=df.dropna()
print("dropping missing value row:\n",dropNa)
print(df.info())

```

1. first check how many rows have null cols
2. then droped that rows which have null values 
3. and 32349 records left.

```python
# Task 3: Business / Operational Data Analysis
# Issue category-wise ticket count
df["Ticket_ID"]
df["Issue_Category"]
```

```python
#● Bar chart (Issue Category vs Ticket Count)
tickett=df["Ticket_ID"].head(8).count()
cattee=df["Issue_Category"].head(8)
#print(tickett)
#print(cattee)
plt.figure(figsize=(10,6))
sns.barplot(x=cattee,y=tickett)
plt.title("Issue Category vs Ticket Count")
plt.xlabel("<------------Category------------>")
plt.xticks(rotation=45)
plt.ylabel("<------------Ticket Counts------------>")
plt.show()
```

```python
r=df["Resolution_Time_Hours"].head(10)
#print(r)
d=df["Created_Date"].head(6)
#print(d)
plt.figure(figsize=(10,6))
sns.lineplot(x=d,y=r)
plt.title("Resolution time")
plt.xlabel("<-----------Date------------->")
plt.xticks(rotation=45)
plt.ylabel("<------------time Hours------------>")
plt.show()
```

```python
cty=df["City"].head(8)
#print(cty)
pri=df["Priority"].head(8)
#print(pri)
plt.figure(figsize=(10,6))
sns.histplot(x=cty,y=pri)
plt.title("(Resolution Time distribution)")
plt.xlabel("<-----------city------------->")
plt.xticks(rotation=45)
plt.ylabel("<------------Priority------------>")
plt.show()
```

```python
typec=df["Satisfaction_Score"].head(8)
#print(typec)
cty1=df["City"].head(8)
#print(cty)
plt.figure(figsize=(10,6))
sns.scatterplot(y=cty1,x=typec)
plt.title("(Resolution Time distribution)")
plt.xlabel("<-----------Satisfy Rate------------->")
plt.xticks(rotation=45)
plt.ylabel("<------------city------------>")
plt.show()
```
